```python
def normalize_video_name(value):
    if pd.isna(value):
        return value

    return str(value).replace(
        "dog_eat_drink_",
        "dog_eat_"
    )
df_mhi["path"] = df_mhi["path"].apply(normalize_video_name) 
df_hog["path"] = df_hog["path"].apply(normalize_video_name) 
df_flow["path"] = df_flow["path"].apply(normalize_video_name) 
```

```python
import os 
def get_filename(path):
    if pd.isna(path):
        return None

    path = str(path).strip()

    # Chuẩn hóa dấu phân cách đường dẫn
    path = path.replace("\\", "/")

    return os.path.basename(path)
```

```python
from itertools import combinations

datasets = {
    "HSV": df_hsv,
    "GLCM": df_glcm,
    "HOG": df_hog,
    "MHI": df_mhi,
    "FLOW": df_flow, 
    "METADATA": df_metadata
}

for (name1, df1), (name2, df2) in combinations(datasets.items(), 2):

    set1 = set(df1["filename"].dropna().unique())
    set2 = set(df2["filename"].dropna().unique())

    only_1 = set1 - set2
    only_2 = set2 - set1

    print(f"\n{'=' * 60}")
    print(f"{name1} so với {name2}")
    print(f"{'=' * 60}")

    print(f"\nCó trong {name1} nhưng không có trong {name2}: {len(only_1)}")
    for filename in sorted(only_1):
        print(" ", filename)

    print(f"\nCó trong {name2} nhưng không có trong {name1}: {len(only_2)}")
    for filename in sorted(only_2):
        print(" ", filename)
```

```python
from pathlib import Path 
import cv2 
def extract_metadata(video_path) -> dict: 
        video_path = Path(video_path)
        cap = cv2.VideoCapture(str(video_path)) 
        if not cap.isOpened(): 
            print(f"[Error]: {video_path}")
            return None  
        metadata = {} 
        metadata["uuid"] = video_path.stem 
        metadata["filename"] = video_path.name 
        metadata["filepath"] = str(video_path) 
        metadata["category"] = video_path.parent.name 
        metadata["fps"] = cap.get(cv2.CAP_PROP_FPS) 
        metadata["total_frames"] = int(cap.get(cv2.CAP_PROP_FRAME_COUNT)) 
        metadata["duration"] = metadata["total_frames"] / metadata["fps"]
        metadata["width"] = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
        metadata["height"] = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
        cap.release() 
        return metadata
mtdt = extract_metadata("/home/ninhdao/D23_Lop4_Nhom16/data_collection/swimming/dog_swimming_41.mp4") 
print(mtdt) 
```
