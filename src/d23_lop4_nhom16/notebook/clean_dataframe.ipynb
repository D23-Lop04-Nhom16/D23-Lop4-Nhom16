```python
import pandas as pd 
```

```python
df_metadata = pd.read_csv("/home/ninhdao/D23_Lop4_Nhom16/src/d23_lop4_nhom16/postgres/metadata.csv", skipinitialspace=True) 
df_flow = pd.read_csv("/home/ninhdao/D23_Lop4_Nhom16/src/d23_lop4_nhom16/postgres/flow_features.csv", skipinitialspace=True) 
df_kff = pd.read_csv("/home/ninhdao/D23_Lop4_Nhom16/src/d23_lop4_nhom16/postgres/keyframe_features.csv", skipinitialspace=True) 
df_kf = pd.read_csv("/home/ninhdao/D23_Lop4_Nhom16/src/d23_lop4_nhom16/postgres/keyframes.csv", skipinitialspace=True) 
df_mhi = pd.read_csv("/home/ninhdao/D23_Lop4_Nhom16/src/d23_lop4_nhom16/postgres/mhi_features.csv", skipinitialspace=True) 
```

```python
def clean_dataframe(df: pd.DataFrame) -> pd.DataFrame: 
    df.columns = [str(col).strip() for col in df.columns]
    for col in df.select_dtypes(include="object").columns:
        df[col] = df[col].astype(str).str.strip()
    return df
```

```python
df_metadata = clean_dataframe(df_metadata) 
df_flow = clean_dataframe(df_flow) 
df_kff = clean_dataframe(df_kff) 
df_kf = clean_dataframe(df_kf) 
df_mhi = clean_dataframe(df_mhi) 
```

```python
df_metadata.to_csv("/home/ninhdao/D23_Lop4_Nhom16/src/d23_lop4_nhom16/postgres/metadata.csv", index=False) 
df_flow.to_csv("/home/ninhdao/D23_Lop4_Nhom16/src/d23_lop4_nhom16/postgres/flow_features.csv", index=False) 
df_kff.to_csv("/home/ninhdao/D23_Lop4_Nhom16/src/d23_lop4_nhom16/postgres/keyframe_features.csv", index=False) 
df_kf.to_csv("/home/ninhdao/D23_Lop4_Nhom16/src/d23_lop4_nhom16/postgres/keyframes.csv", index=False) 
df_mhi.to_csv("/home/ninhdao/D23_Lop4_Nhom16/src/d23_lop4_nhom16/postgres/mhi_features.csv", index=False) 
```
