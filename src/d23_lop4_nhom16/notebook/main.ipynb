### Color histogram

```python
import cv2 
import numpy as np 
import matplotlib.pyplot as plt

def extract_hsv_histogram(frame, h_bins=8, s_bins=3, v_bins=3): 
    hsv = cv2.cvtColor(frame, cv2.COLOR_BGR2HSV).astype(np.int32)  

    h = hsv[:,:,0]
    s = hsv[:,:,1]
    v = hsv[:,:,2] 

    h_idx = (h * h_bins / 180).astype(int) 
    s_idx = (s * s_bins / 256).astype(int)
    v_idx = (v * v_bins / 256).astype(int)

    h_idx = np.clip(h_idx, 0, h_bins-1) 
    s_idx = np.clip(s_idx, 0, s_bins-1)
    v_idx = np.clip(v_idx, 0, v_bins-1) 

    bin_idx = h_idx * (s_bins * v_bins) + s_idx * v_bins + v_idx 

    hist = np.bincount(bin_idx.ravel(), minlength = h_bins * s_bins * v_bins) 

    hist = hist.astype(float) 

    if hist.sum() > 0:
        hist /= hist.sum() 

    return hist 
```

```python
def extract_video_hsv_histograms(video_path): 
    cap = cv2.VideoCapture(video_path) 
    if not cap.isOpened(): 
        print(f"Không mở được video: {video_path}") 
        return None
    fps = cap.get(cv2.CAP_PROP_FPS) 
    if fps <= 0: 
        fps = 25  
    frame_interval = int(round(fps)) 
    keyframe_features = [] 
    keyframe_images = [] 
    frame_idx = 0 
    keyframe_count = 0 

    while True: 
        ret, frame = cap.read() 
        if not ret: 
            break 
        if frame_idx % frame_interval == 0: 
            hist = extract_hsv_histogram(frame) 
            keyframe_features.append(hist) 
            keyframe_images.append(frame) 
            keyframe_count += 1
        frame_idx += 1
    cap.release() 
    features = np.array(keyframe_features, dtype=np.float32) 
    return features, keyframe_images 
```

### GLCM features

```python
import cv2 
import numpy as np 
from math import log2 
```

```python
def quantize_gray(gray, levels=8): 
    return (gray.astype(np.float32) / 256 * levels).astype(np.int32) 
```

```python
def compute_glcm(gray_q, levels=8, dx=1, dy=0): 
    h, w = gray_q.shape 
    glcm = np.zeros((levels, levels), dtype=np.float64) 

    for i in range(h): 
        for j in range(w): 
            x2 = i+dy 
            y2 = j+dx 
            if 0<=x2<h and 0<=y2<w: 
                g1 = gray_q[i, j]
                g2 = gray_q[x2, y2]
                glcm[g1, g2] += 1
    
    total = glcm.sum()
    if total > 0:
        glcm /= total
    return glcm
```

```python
def glcm_features(glcm):
    """
    Tính 4 đặc trưng từ một ma trận GLCM:
    Energy, Contrast, Entropy, Homogeneity
    """
    levels = glcm.shape[0]
    energy = 0.0
    contrast = 0.0
    entropy = 0.0
    homogeneity = 0.0

    for i in range(levels):
        for j in range(levels):
            p = glcm[i, j]
            energy += p ** 2
            contrast += ((i - j) ** 2) * p
            if p > 0:
                entropy -= p * log2(p)
            homogeneity += p / (1 + abs(i - j))

    return np.array([energy, contrast, entropy, homogeneity], dtype=np.float32)
```

```python
def extract_glcm_feature(frame, levels=8):
    """
    Trích xuất vector GLCM 4 chiều từ một frame (keyframe)
    đúng theo mô tả trong báo cáo.
    """
    # 1. Chuyển sang ảnh xám
    if len(frame.shape) == 3:
        gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
    else:
        gray = frame

    # 2. Lượng tử hóa về L=8 mức xám
    gray_q = quantize_gray(gray, levels=levels)

    # 3. 4 hướng với d=1
    # (dx, dy): 0°, 45°, 90°, 135°
    directions = [ 
        (1, 0),   # 0 độ   (ngang)
        (1, 1),   # 45 độ
        (0, 1),   # 90 độ  (dọc)
        (-1, 1)   # 135 độ
    ]

    features_list = []
    for dx, dy in directions:
        glcm = compute_glcm(gray_q, levels=levels, dx=dx, dy=dy)
        feat = glcm_features(glcm)
        features_list.append(feat)

    # 4. Lấy trung bình trên 4 hướng → vector 4 chiều
    feature_vector = np.mean(features_list, axis=0)
    return feature_vector
```

```python
def extract_glcm_from_video(video_path):
    """
    Lấy 1 frame mỗi giây và trích GLCM feature
    Trả về: mảng shape (K, 4)
    """
    cap = cv2.VideoCapture(video_path)
    if not cap.isOpened():
        raise ValueError(f"Không mở được video: {video_path}")

    fps = cap.get(cv2.CAP_PROP_FPS)
    if fps <= 0:
        fps = 25

    frame_interval = int(round(fps))  # 1 giây
    features = []
    frame_idx = 0
    frame_images = []

    while True:
        ret, frame = cap.read()
        if not ret:
            break
        
        if frame_idx % frame_interval == 0:
            vec = extract_glcm_feature(frame)   # shape (4,)
            frame_images.append(frame)  
            features.append(vec)

        frame_idx += 1

    cap.release()
    return frame_images, np.array(features, dtype=np.float32)  # shape (K, 4)
```

```python
import numpy as np, cv2, matplotlib.pyplot as plt

DIRS = [(1, 0), (1, 1), (0, 1), (-1, 1)]
NAMES = ["energy", "contrast", "entropy", "homogeneity"]

def to_gray_q(frame, levels=8):
    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY) if frame.ndim == 3 else frame
    return quantize_gray(gray, levels=levels)   # quantize CẢ frame 1 lần

def avg_glcm(gray_q, levels=8):
    """Ma trận GLCM trung bình 4 hướng (để vẽ)."""
    return np.mean([compute_glcm(gray_q, levels=levels, dx=dx, dy=dy)
                    for dx, dy in DIRS], axis=0)

def local_feature_map(gray_q, levels=8, win=48, step=24):
    """Tính đúng feature của bạn trên từng cửa sổ trượt."""
    H, W = gray_q.shape
    rows, cols = (H - win) // step + 1, (W - win) // step + 1
    out = np.zeros((4, rows, cols))
    for r in range(rows):
        for c in range(cols):
            p = gray_q[r*step:r*step+win, c*step:c*step+win]
            f = np.mean([glcm_features(compute_glcm(p, levels=levels, dx=dx, dy=dy))
                         for dx, dy in DIRS], axis=0)
            out[:, r, c] = f
    return out

def visualize_frame(frame, levels=8, win=48, step=24):
    gray_q = to_gray_q(frame, levels)
    g = avg_glcm(gray_q, levels)
    g = g / g.sum()
    vec = extract_glcm_feature(frame, levels)
    maps = local_feature_map(gray_q, levels, win, step)

    fig = plt.figure(figsize=(20, 8))
    # Hàng 1: frame, ảnh sau lượng tử hóa, ma trận GLCM, vector
    ax = fig.add_subplot(2, 4, 1)
    ax.imshow(cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)); ax.set_title("frame"); ax.axis("off")
    ax = fig.add_subplot(2, 4, 2)
    ax.imshow(gray_q, cmap="gray", vmin=0, vmax=levels-1); ax.set_title(f"sau lượng tử hóa L={levels}"); ax.axis("off")
    ax = fig.add_subplot(2, 4, 3)
    im = ax.imshow(np.log1p(g * 1e4), cmap="magma")   # log để thấy rõ phần nhỏ
    ax.set_title("GLCM trung bình 4 hướng (log)"); ax.set_xlabel("mức xám j"); ax.set_ylabel("mức xám i")
    plt.colorbar(im, ax=ax, fraction=0.046)
    ax = fig.add_subplot(2, 4, 4)
    ax.bar(NAMES, vec); ax.set_title("vector 4 chiều"); ax.tick_params(axis="x", rotation=30)

    # Hàng 2: heatmap cục bộ 4 feature, resize về kích thước frame để chồng lên ảnh
    H, W = gray_q.shape
    base = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB) if frame.ndim == 3 else frame
    for i in range(4):
        ax = fig.add_subplot(2, 4, 5 + i)
        hm = cv2.resize(maps[i], (W, H), interpolation=cv2.INTER_LINEAR)
        ax.imshow(base, cmap="gray")
        im = ax.imshow(hm, cmap="jet", alpha=0.55)
        ax.set_title(f"{NAMES[i]} (toàn frame = {vec[i]:.3f})"); ax.axis("off")
        plt.colorbar(im, ax=ax, fraction=0.046)
    plt.tight_layout(); plt.show()

visualize_frame(frame_images[0])
```

```python
import pickle
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

from d23_lop4_nhom16.config import GLCM_FEATURE_PATH


# ==========================================
# 1. LOAD GLCM FEATURES
# ==========================================

def load_glcm_features(pkl_path):
    with open(pkl_path, "rb") as f:
        records = pickle.load(f)

    data = []

    for record in records:
        feature = np.asarray(
            record["feature"],
            dtype=np.float32
        )

        if feature.shape != (4,):
            print("Bo qua vector khong hop le:", record)
            continue

        if not np.all(np.isfinite(feature)):
            print("Bo qua vector co NaN/Inf:", record)
            continue

        video_path = record["video_path"]

        # Mặc định lấy tên thư mục cha làm nhãn hành động
        action = Path(video_path).parent.name

        data.append({
            "video_id": record.get("video_id"),
            "video_name": record["video_name"],
            "video_path": video_path,
            "frame_idx": record["frame_idx"],
            "timestamp_sec": record["timestamp_sec"],
            "action": action,
            "Energy": feature[0],
            "Contrast": feature[1],
            "Entropy": feature[2],
            "Homogeneity": feature[3]
        })

    return pd.DataFrame(data)


# ==========================================
# 2. VISUALIZE DISTRIBUTION
# ==========================================

def visualize_histogram(df, output_dir):

    features = [
        "Energy",
        "Contrast",
        "Entropy",
        "Homogeneity"
    ]

    fig, axes = plt.subplots(2, 2, figsize=(12, 8))

    for ax, feature in zip(axes.flatten(), features):

        ax.hist(
            df[feature],
            bins=40,
            edgecolor="black"
        )

        ax.set_title(f"Distribution of {feature}")
        ax.set_xlabel(feature)
        ax.set_ylabel("Frequency")

    plt.tight_layout()

    plt.savefig(
        output_dir / "glcm_histogram.png",
        dpi=300
    )

    plt.show()


# ==========================================
# 3. VISUALIZE BOX PLOT BY ACTION
# ==========================================

def visualize_boxplot(df, output_dir):

    features = [
        "Energy",
        "Contrast",
        "Entropy",
        "Homogeneity"
    ]

    actions = sorted(df["action"].unique())

    fig, axes = plt.subplots(
        2, 2,
        figsize=(16, 10)
    )

    for ax, feature in zip(axes.flatten(), features):

        values = [
            df.loc[df["action"] == action, feature].values
            for action in actions
        ]

        ax.boxplot(
            values,
            tick_labels=actions,
            showfliers=False
        )

        ax.set_title(f"{feature} by Dog Action")
        ax.set_ylabel(feature)
        ax.tick_params(axis="x", rotation=45)

    plt.tight_layout()

    plt.savefig(
        output_dir / "glcm_boxplot.png",
        dpi=300
    )

    plt.show()


# ==========================================
# 4. PCA VISUALIZATION
# ==========================================

def visualize_pca(df, output_dir):

    features = [
        "Energy",
        "Contrast",
        "Entropy",
        "Homogeneity"
    ]

    X = df[features].values

    # Standardize 4 GLCM dimensions
    scaler = StandardScaler()
    X_scaled = scaler.fit_transform(X)

    # Reduce 4 dimensions to 2 dimensions
    pca = PCA(n_components=2)

    X_pca = pca.fit_transform(X_scaled)

    df = df.copy()

    df["PC1"] = X_pca[:, 0]
    df["PC2"] = X_pca[:, 1]

    # Plot
    plt.figure(figsize=(12, 9))

    actions = sorted(df["action"].unique())

    for action in actions:

        subset = df[df["action"] == action]

        plt.scatter(
            subset["PC1"],
            subset["PC2"],
            label=action,
            alpha=0.5,
            s=15
        )

    plt.title("PCA Visualization of GLCM Features")
    plt.xlabel("Principal Component 1")
    plt.ylabel("Principal Component 2")

    plt.legend(
        bbox_to_anchor=(1.05, 1),
        loc="upper left"
    )

    plt.tight_layout()

    plt.savefig(
        output_dir / "glcm_pca.png",
        dpi=300
    )

    plt.show()

    print("\nPCA Explained Variance:")
    print(pca.explained_variance_ratio_)

    print(
        "Total explained variance:",
        pca.explained_variance_ratio_.sum()
    )


# ==========================================
# 5. MAIN
# ==========================================

def main():

    output_dir = Path("glcm_visualization")
    output_dir.mkdir(exist_ok=True)

    df = load_glcm_features(GLCM_FEATURE_PATH)

    if df.empty:
        print("Khong co GLCM features!")
        return

    print("Total keyframes:", len(df))
    print("Total videos:", df["video_id"].nunique())

    print("\nNumber of keyframes by action:")
    print(df.groupby("action").size())

    print("\nGLCM statistics:")
    print(
        df[[
            "Energy",
            "Contrast",
            "Entropy",
            "Homogeneity"
        ]].describe()
    )

    visualize_histogram(df, output_dir)

    visualize_boxplot(df, output_dir)

    visualize_pca(df, output_dir)

    print("\nDa hoan thanh truc quan hoa GLCM!")
    print("Ket qua duoc luu tai:", output_dir)


if __name__ == "__main__":
    main()
```

```python
# ==========================================
# 1. LOAD HSV FEATURES
# ==========================================

def load_hsv_features(feature_path):
    with open(feature_path, "rb") as f:
        all_features = pickle.load(f)
    # Gom các keyframe theo video_path
    video_features = defaultdict(list)
    for item in all_features:
        video_path = item["video_path"]
        feature = item["feature"]
        video_features[video_path].append(feature)
    # Chuyển danh sách feature thành numpy array
    for video_path in video_features:
        video_features[video_path] = np.array(
            video_features[video_path],
            dtype=np.float32
        )
    return video_features


# ==========================================
# 2. TINH DO TUONG DONG GIUA 2 HISTOGRAM
# ==========================================

def histogram_similarity(hist1, hist2):
    """
    Histogram Intersection.
    Giá trị nằm trong [0, 1].
    Càng gần 1: hai histogram càng giống nhau.
    """
    return np.minimum(hist1, hist2).sum()


# ==========================================
# 3. TINH DO TUONG DONG GIUA 2 VIDEO
# ==========================================

def video_similarity(query_features, candidate_features):
    """
    So sánh hai tập hợp HSV Histogram.

    Với mỗi keyframe của video A:
        - Tìm keyframe giống nhất trong video B.
        - Tính trung bình độ tương đồng.

    Thực hiện tương tự theo chiều ngược lại.
    Cuối cùng lấy trung bình hai chiều.
    """

    # A -> B
    similarities_ab = []

    for query_hist in query_features:

        scores = np.minimum(
            candidate_features,
            query_hist
        ).sum(axis=1)

        similarities_ab.append(scores.max())

    score_ab = np.mean(similarities_ab)

    # B -> A
    similarities_ba = []

    for candidate_hist in candidate_features:

        scores = np.minimum(
            query_features,
            candidate_hist
        ).sum(axis=1)

        similarities_ba.append(scores.max())

    score_ba = np.mean(similarities_ba)

    # Độ tương đồng đối xứng
    return (score_ab + score_ba) / 2


# ==========================================
# 4. TIM TOP 5 VIDEO TUONG DONG
# ==========================================

def retrieve_similar_videos(
    query_features,
    database_features,
    top_k=5,
    exclude_path=None
):

    results = []

    for video_path, candidate_features in database_features.items():

        # Không so sánh video với chính nó nếu được yêu cầu
        if exclude_path is not None:

            if Path(video_path).resolve() == Path(exclude_path).resolve():
                continue

        similarity = video_similarity(
            query_features,
            candidate_features
        )

        results.append({
            "video_path": video_path,
            "similarity": similarity
        })

    # Sắp xếp giảm dần theo độ tương đồng
    results.sort(
        key=lambda x: x["similarity"],
        reverse=True
    )

    return results[:top_k]


# ==========================================
# 5. MAIN
# ==========================================

if __name__ == "__main__":

    # Nhập video truy vấn
    query_video = input("Nhap duong dan video query: ").strip()

    query_video = Path(query_video).expanduser()

    if not query_video.is_file():
        raise FileNotFoundError(
            f"Khong tim thay video: {query_video}"
        )

    # Trích xuất HSV Histogram của video query
    print("\nDang trich xuat HSV Histogram...")

    query_features = extract_video_hsv_histogram(
        str(query_video)
    )

    if not query_features:
        raise ValueError("Khong trich xuat duoc feature tu video query.")

    query_features = np.array(
        [item["feature"] for item in query_features],
        dtype=np.float32
    )

    print(f"So keyframe query: {len(query_features)}")

    # Load database HSV Histogram
    print("\nDang load HSV database...")

    database_features = load_hsv_features(
        HSV_FEATURE_PATH
    )

    print(f"Tong so video trong database: {len(database_features)}")

    # Tìm Top 5 video tương tự
    print("\nDang tim kiem video tuong tu...")

    results = retrieve_similar_videos(
        query_features,
        database_features,
        top_k=5,
        exclude_path=query_video
    )

    # Hiển thị kết quả
    print("\n" + "=" * 70)
    print("TOP 5 VIDEO TUONG DONG VOI QUERY VIDEO")
    print("=" * 70)

    for rank, result in enumerate(results, start=1):

        print(f"\nRank: {rank}")
        print(f"Video: {result['video_path']}")
        print(
            f"Similarity: {result['similarity']:.4f} "
            f"({result['similarity'] * 100:.2f}%)"
        )

    print("\n" + "=" * 70)
```

```python
import pickle
from pathlib import Path

import cv2
import numpy as np

from d23_lop4_nhom16.config import (
    HSV_FEATURE_PATH,
    GLCM_FEATURE_PATH
)

from d23_lop4_nhom16.feature_extraction.color_histogram import (
    extract_video_hsv_histogram, 
    extract_hsv_histogram 
)

from d23_lop4_nhom16.feature_extraction.glcm import (
    extract_glcm_from_video, 
    extract_glcm_feature 
)


# =====================================================
# 1. CONFIGURATION
# =====================================================

TOP_K = 5

W_HSV = 0.5
W_GLCM = 0.5

EPSILON = 1e-8


# =====================================================
# 2. LOAD DATABASE FEATURES
# =====================================================

def normalize_path(video_path):
    """
    Chuẩn hóa đường dẫn để HSV và GLCM
    có thể nhóm đúng các keyframe thuộc cùng video.
    """
    return str(Path(video_path).expanduser().resolve())


def load_pickle(path):

    with open(path, "rb") as f:
        return pickle.load(f)


def group_features(records, feature_name):
    """
    Nhóm các vector keyframe theo video_path.

    Output:
    {
        video_path_1: [
            feature_frame_1,
            feature_frame_2,
            ...
        ],
        video_path_2: [...]
    }
    """

    grouped = {}

    for record in records:

        path = normalize_path(record["video_path"])

        feature = np.asarray(
            record["feature"],
            dtype=np.float32
        )

        if not np.all(np.isfinite(feature)):
            continue

        grouped.setdefault(path, []).append(feature)

    # Chuyển list thành numpy array
    for path in grouped:
        grouped[path] = np.asarray(
            grouped[path],
            dtype=np.float32
        )

    print(
        f"{feature_name}: "
        f"{len(grouped)} videos loaded"
    )

    return grouped


def load_database():

    hsv_records = load_pickle(HSV_FEATURE_PATH)
    glcm_records = load_pickle(GLCM_FEATURE_PATH)

    hsv_db = group_features(
        hsv_records,
        "HSV"
    )

    glcm_db = group_features(
        glcm_records,
        "GLCM"
    )

    # Chỉ lấy các video có đủ cả HSV và GLCM
    common_paths = set(hsv_db) & set(glcm_db)

    hsv_db = {
        path: hsv_db[path]
        for path in common_paths
    }

    glcm_db = {
        path: glcm_db[path]
        for path in common_paths
    }

    print(f"\nTotal videos available: {len(common_paths)}")

    if not common_paths:
        raise ValueError(
            "Khong tim thay video chung trong HSV va GLCM database. "
            "Hay kiem tra lai video_path."
        )

    return hsv_db, glcm_db


# =====================================================
# 3. CALCULATE GLCM SCALER
# =====================================================

def calculate_glcm_scaler(glcm_db):

    all_features = np.concatenate(
        list(glcm_db.values()),
        axis=0
    )

    mean = np.mean(all_features, axis=0)
    std = np.std(all_features, axis=0)

    std[std < EPSILON] = 1.0

    print("\nGLCM Scaler:")
    print("Mean:", mean)
    print("Std :", std)

    return mean, std


def standardize_glcm(features, mean, std):

    return (features - mean) / std


# =====================================================
# 4. HSV SIMILARITY
# =====================================================

def histogram_intersection(a, b):

    return np.minimum(a, b).sum(axis=-1)


def calculate_hsv_similarity(A, B):
    """
    Symmetric nearest-neighbor Histogram Intersection.

    A: (N, 72)
    B: (M, 72)
    """

    # Ma trận similarity giữa các keyframe
    similarity_matrix = np.minimum(
        A[:, None, :],
        B[None, :, :]
    ).sum(axis=2)

    # Mỗi keyframe của A tìm keyframe tương đồng nhất trong B
    a_to_b = similarity_matrix.max(axis=1).mean()

    # Mỗi keyframe của B tìm keyframe tương đồng nhất trong A
    b_to_a = similarity_matrix.max(axis=0).mean()

    return float((a_to_b + b_to_a) / 2)


# =====================================================
# 5. GLCM DISTANCE
# =====================================================

def calculate_glcm_distance(A, B):
    """
    Symmetric Chamfer Distance.

    A: (N, 4) - standardized GLCM
    B: (M, 4) - standardized GLCM
    """

    # Tính khoảng cách Euclidean giữa các keyframe
    distance_matrix = np.linalg.norm(
        A[:, None, :] - B[None, :, :],
        axis=2
    )

    # Mỗi keyframe A tìm keyframe gần nhất trong B
    a_to_b = distance_matrix.min(axis=1).mean()

    # Mỗi keyframe B tìm keyframe gần nhất trong A
    b_to_a = distance_matrix.min(axis=0).mean()

    return float((a_to_b + b_to_a) / 2)


# =====================================================
# 6. EXTRACT QUERY FEATURES
# =====================================================

def extract_query_features(query_video_path):

    query_video_path = normalize_path(query_video_path)

    if not Path(query_video_path).is_file():
        raise FileNotFoundError(
            f"Khong tim thay video: {query_video_path}"
        )

    print("\nExtracting HSV features...")

    query_hsv_records = extract_video_hsv_histogram(
        query_video_path
    )

    print("Extracting GLCM features...")

    query_glcm_records = extract_glcm_from_video(
        query_video_path
    )

    if not query_hsv_records or not query_glcm_records:
        raise ValueError(
            "Khong trich xuat duoc features tu video truy van."
        )

    query_hsv = np.asarray(
        [r["feature"] for r in query_hsv_records],
        dtype=np.float32
    )

    query_glcm = np.asarray(
        [r["feature"] for r in query_glcm_records],
        dtype=np.float32
    )

    if query_hsv.ndim != 2 or query_hsv.shape[1] != 72:
        raise ValueError(
            f"HSV feature khong hop le: {query_hsv.shape}"
        )

    if query_glcm.ndim != 2 or query_glcm.shape[1] != 4:
        raise ValueError(
            f"GLCM feature khong hop le: {query_glcm.shape}"
        )

    return query_video_path, query_hsv, query_glcm


# =====================================================
# 7. COMBINED RETRIEVAL
# =====================================================

def retrieve_top_k(
    query_video_path,
    hsv_db,
    glcm_db,
    mean,
    std,
    top_k=TOP_K
):

    query_path, query_hsv, query_glcm = (
        extract_query_features(query_video_path)
    )

    # Chuẩn hóa GLCM query bằng scaler của database
    query_glcm_scaled = standardize_glcm(
        query_glcm,
        mean,
        std
    )

    # Chuẩn hóa GLCM database trong RAM
    glcm_db_scaled = {
        path: standardize_glcm(features, mean, std)
        for path, features in glcm_db.items()
    }

    results = []

    print("\nCalculating similarities...")

    # Bỏ qua chính video truy vấn nếu nó đã có trong database
    candidate_paths = [
        path for path in hsv_db
        if path != query_path
    ]

    if not candidate_paths:
        raise ValueError("Khong co video nao de so sanh.")

    for idx, candidate_path in enumerate(candidate_paths):

        candidate_hsv = hsv_db[candidate_path]
        candidate_glcm = glcm_db_scaled[candidate_path]

        # HSV similarity
        hsv_similarity = calculate_hsv_similarity(
            query_hsv,
            candidate_hsv
        )

        # GLCM distance
        glcm_distance = calculate_glcm_distance(
            query_glcm_scaled,
            candidate_glcm
        )

        results.append({
            "video_path": candidate_path,
            "hsv_similarity": hsv_similarity,
            "glcm_distance": glcm_distance
        })

        if (idx + 1) % 50 == 0:
            print(
                f"Compared: {idx + 1}/{len(candidate_paths)}"
            )

    # -------------------------------------------------
    # Convert GLCM distance to similarity
    # -------------------------------------------------

    distances = np.array(
        [r["glcm_distance"] for r in results]
    )

    # Median distance của các ứng viên trong lần truy vấn này
    tau = np.median(distances)

    if tau < EPSILON:
        tau = 1.0

    for result in results:

        glcm_similarity = np.exp(
            -result["glcm_distance"] / tau
        )

        result["glcm_similarity"] = float(
            glcm_similarity
        )

        # Combined similarity
        result["combined_similarity"] = (
            W_HSV * result["hsv_similarity"]
            + W_GLCM * glcm_similarity
        )

    # Sắp xếp giảm dần theo combined similarity
    results.sort(
        key=lambda x: x["combined_similarity"],
        reverse=True
    )

    return results[:top_k]


# =====================================================
# 8. DISPLAY RESULTS
# =====================================================

def display_results(results):

    print("\n" + "=" * 90)
    print("TOP 5 MOST SIMILAR VIDEOS")
    print("=" * 90)

    for rank, result in enumerate(results, start=1):

        print(f"\nRank {rank}")

        print(
            "Video:",
            Path(result["video_path"]).name
        )

        print(
            "Path:",
            result["video_path"]
        )

        print(
            f"HSV Similarity: "
            f"{result['hsv_similarity']:.6f}"
        )

        print(
            f"GLCM Distance: "
            f"{result['glcm_distance']:.6f}"
        )

        print(
            f"GLCM Similarity: "
            f"{result['glcm_similarity']:.6f}"
        )

        print(
            f"Combined Similarity: "
            f"{result['combined_similarity']:.6f}"
        )

        print("-" * 90)


# =====================================================
# 9. MAIN
# =====================================================

def main():

    print("=" * 60)
    print("VIDEO RETRIEVAL - HSV + GLCM")
    print("=" * 60)

    query_video_path = input(
        "\nNhap duong dan video truy van: "
    ).strip().strip('"').strip("'")

    # Load database
    hsv_db, glcm_db = load_database()

    # Tính scaler từ GLCM database
    mean, std = calculate_glcm_scaler(glcm_db)

    # Retrieve Top-K
    results = retrieve_top_k(
        query_video_path=query_video_path,
        hsv_db=hsv_db,
        glcm_db=glcm_db,
        mean=mean,
        std=std,
        top_k=TOP_K
    )

    # Display
    display_results(results)


if __name__ == "__main__":
    main()
```

```python
"""Run all project feature extractors on one video and print their results."""

from __future__ import annotations

import json
from pathlib import Path
from typing import Any

import cv2
import numpy as np

from d23_lop4_nhom16.feature_extraction.color_histogram import (
    extract_video_hsv_histogram,
)
from d23_lop4_nhom16.feature_extraction.flow_extracter import extract_flow
from d23_lop4_nhom16.feature_extraction.glcm import extract_glcm_from_video
from d23_lop4_nhom16.feature_extraction.hog_extracter import extract_video_hog
from d23_lop4_nhom16.feature_extraction.mhi_extracter import extract_mhi


class NumpyJSONEncoder(json.JSONEncoder):
    """Convert NumPy values to standard JSON-compatible values."""

    def default(self, value: Any) -> Any:
        if isinstance(value, np.ndarray):
            return value.tolist()
        if isinstance(value, np.generic):
            return value.item()
        if isinstance(value, Path):
            return str(value)
        return super().default(value)


def extract_video_features(video_path: Path) -> dict[str, Any]:
    """Extract HOG, optical flow, GLCM, HSV, and MHI features for one video."""
    video_path = video_path.expanduser().resolve()
    if not video_path.is_file():
        raise FileNotFoundError(f"Không tìm thấy video: {video_path}")

    flow_result = extract_flow(video_path)
    hog_result = extract_video_hog(video_path)
    glcm_records = extract_glcm_from_video(str(video_path))
    hsv_records = extract_video_hsv_histogram(str(video_path))
    if hsv_records is None:
        raise ValueError(f"Không mở được video để trích HSV: {video_path}")
    mhi_result = extract_mhi(video_path)

    metadata = {
        "filename": video_path.name,
        "category": video_path.parent.name,
        **flow_result["metadata"],
    }
    return {
        "video": metadata,
        "features": {
            "hog": hog_result,
            "flow": flow_result["flow_vector"],
            "glcm": glcm_records,
            "hsv": hsv_records,
            "mhi": mhi_result["mhi_vectors"],
        },
    }


def main() -> None:
    # Nhập đường dẫn video từ bàn phím
    raw = input("Nhập đường dẫn video: ").strip()
    # Bỏ dấu ngoặc kép/đơn nếu người dùng copy-paste kèm
    raw = raw.strip('"').strip("'")

    if not raw:
        print("Chưa nhập đường dẫn.")
        return

    video_path = Path(raw)

    try:
        result = extract_video_features(video_path)
    except (OSError, ValueError, cv2.error) as error:
        print(f"Lỗi: {error}")
        return

    # In full JSON
    output = json.dumps(
        result,
        cls=NumpyJSONEncoder,
        ensure_ascii=False,
        indent=2,
    )
    print(output)

    # Tóm tắt shape từng feature
    print("\n--- Tóm tắt shape ---")
    feats = result["features"]
    print(f"HOG   : {np.asarray(feats['hog']['hog']).shape}")
    print(f"Flow  : {np.asarray(feats['flow']).shape}")
    print(f"GLCM  : {len(feats['glcm'])} keyframe x "
          f"{len(feats['glcm'][0]['feature'])} chiều")
    print(f"HSV   : {len(feats['hsv'])} keyframe x "
          f"{len(feats['hsv'][0]['feature'])} chiều")
    print(f"MHI   : {np.asarray(feats['mhi']).shape}")


if __name__ == "__main__":
    main()
```
