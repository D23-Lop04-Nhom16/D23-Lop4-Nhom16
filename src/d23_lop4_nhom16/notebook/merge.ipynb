```python
import pandas as pd

df_merged = (
    df_hsv.rename(columns={"vector": "hsv_vector"})
    .merge(
        df_glcm.rename(columns={"vector": "glcm_vector"}),
        on="keyframe_id",
        how="inner",
    )
    .merge(
        df_hog.rename(columns={"vector": "hog_vector"}),
        on="keyframe_id",
        how="inner",
    )
    .sort_values("keyframe_id")
    .reset_index(drop=True)
)

display(df_merged.head())
print(df_merged.shape)
```

```python
import pandas as pd

# =====================================================
# 1. ĐỒNG BỘ keyframe_id (df_merge: 0-based, df_keyframe: 1-based)
# =====================================================
print("Trước đồng bộ:")
print(f"  df_merge    keyframe_id: {df_merged['keyframe_id'].min()} -> {df_merged['keyframe_id'].max()}")
print(f"  df_keyframe keyframe_id: {df_keyframe['keyframe_id'].min()} -> {df_keyframe['keyframe_id'].max()}")

# Đưa df_keyframe về 0-based cho khớp df_merge
df_keyframe["keyframe_id"] = df_keyframe["keyframe_id"] - 1

print("Sau đồng bộ:")
print(f"  df_keyframe keyframe_id: {df_keyframe['keyframe_id'].min()} -> {df_keyframe['keyframe_id'].max()}")

# =====================================================
# 2. MERGE 3 BẢNG (inner — chỉ phần chung)
# =====================================================

# 2a: df_merge + df_keyframe -> có video_id
df_step1 = df_merged.merge(
    df_keyframe[["keyframe_id", "video_id", "frame_idx",
                 "timestamp_sec", "keyframe_index"]],
    on="keyframe_id",
    how="inner",
)
print(f"\ndf_merge + df_keyframe (inner): {df_step1.shape}")

# 2b: thêm metadata video
df_full = df_step1.merge(
    df_metadata[["video_id", "filename", "category",
                 "fps", "duration", "total_frames"]],
    on="video_id",
    how="inner",
)
print(f"+ df_metadata (inner): {df_full.shape}")

# =====================================================
# 3. KIỂM TRA PHẦN CHUNG
# =====================================================
v_merge_kf = set(df_step1["video_id"].dropna().astype(int))
v_meta     = set(df_metadata["video_id"].astype(int))

print(f"\n--- Phần chung 3 bảng ---")
print(f"Video trong (df_merge + df_keyframe): {len(v_merge_kf)}")
print(f"Video trong df_metadata            : {len(v_meta)}")
print(f"Video có mặt ở CẢ 3 bảng           : {len(v_merge_kf & v_meta)}")
print(f"Video có trong merge+keyframe nhưng THIẾU metadata: {len(v_merge_kf - v_meta)}")
if v_merge_kf - v_meta:
    print(f"  -> {sorted(v_merge_kf - v_meta)[:20]}")
print(f"Video có trong metadata nhưng THIẾU feature       : {len(v_meta - v_merge_kf)}")
if v_meta - v_merge_kf:
    print(f"  -> {sorted(v_meta - v_merge_kf)[:20]}")

# =====================================================
# 4. SỐ KEYFRAME MỖI VIDEO (chỉ phần chung)
# =====================================================
kf_per_video = (
    df_full
    .groupby(["video_id", "filename", "category"])
    .size()
    .rename("n_keyframes")
    .reset_index()
    .sort_values("video_id")
)

print(f"\n--- Số keyframe/video (phần chung 3 bảng) ---")
print(f"Tổng keyframe : {kf_per_video['n_keyframes'].sum()}")
print(f"Số video      : {len(kf_per_video)}")
print(f"\nThống kê:")
print(kf_per_video["n_keyframes"].describe())

print(f"\n5 video đầu:")
print(kf_per_video.head().to_string(index=False))
print(f"\n5 video cuối:")
print(kf_per_video.tail().to_string(index=False))

# =====================================================
# 5. SO SÁNH VỚI df_keyframe GỐC (phát hiện mất keyframe)
# =====================================================
kf_ref    = df_keyframe.groupby("video_id").size().rename("ref")
kf_merged = df_full.groupby("video_id").size().rename("merged")
cmp = pd.concat([kf_ref, kf_merged], axis=1).fillna(0).astype(int)
cmp["match"] = cmp["ref"] == cmp["merged"]

print(f"\n--- So sánh keyframe gốc vs sau merge ---")
print(f"Video khớp     : {cmp['match'].sum()} / {len(cmp)}")
print(f"Video KHÔNG khớp: {(~cmp['match']).sum()}")
if (~cmp["match"]).any():
    print("\nVideo lệch (ref vs merged):")
    print(cmp[~cmp["match"]].head(20).to_string())
```
