```python
import pandas as pd 
from d23_lop4_nhom16.config import get_db_url 
from sqlalchemy import create_engine 
```

```python
DATABASE_URL = get_db_url() 
engine = create_engine(DATABASE_URL) 

query_keyframes = """
SELECT 
    v.video_id,
    v.filename,
    v.category,
    k.keyframe_id,
    k.frame_idx,
    k.timestamp_sec
FROM videos v
JOIN keyframes k ON v.video_id = k.video_id
WHERE v.category = 'Eating'
ORDER BY v.video_id, k.timestamp_sec;
""" 
df_keyframes = pd.read_sql(query_keyframes, con=engine)
print("Danh sách keyframe thuộc category 'Eating':")
print(df_keyframes.head())
```
